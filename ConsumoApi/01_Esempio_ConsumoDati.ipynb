```python
import matplotlib.pyplot as plt
import pandas as pd

UrlApi = {
    "Mese": ['Gen', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    "Vendite": [1200, 1500, 1300, 1800, 1650, 1340]
}

df = pd.DataFrame(UrlApi)
df
```

# Calcolare una statistica

```python
media_vendite = df['Vendite'].mean()
print(f'La media delle vendite è: {media_vendite:.2f}')
```

# Creazione del grafico

# PLOT

```python
plt.plot(df['Mese'], df['Vendite'], marker='o', color='red')
plt.title('Andamento delle vendite')
plt.xlabel("Mese")
plt.ylabel('Euro()')
plt.grid(True)

plt.show()
```

# HISTOGRAM

```python
plt.hist(df['Vendite'], bins=5, edgecolor='white', color='blue')
plt.title('Istogramma delle vendite')
plt.xlabel("Euro()")
plt.ylabel('Frequenza')
plt.grid(True)

plt.show()
```
